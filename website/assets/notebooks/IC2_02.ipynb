**Begriffe, die immer wieder auftauchen**:

*Data Preprocessing* - Die Aufbereitung von Rohdaten für die Analyse. Umfasst Bereinigung, Transformation, Normalisierung und das Handling von fehlenden Werten.

*Exploratory Data Analysis (EDA)* - Der erste Schritt jeder Datenanalyse. Dabei werden Muster, Anomalien und Zusammenhänge in den Daten entdeckt, meist visuell.

*Feature Engineering* - Das Erstellen neuer, aussagekräftiger Features aus vorhandenen Daten, um die Modellperformance zu verbessern.

*Data Visualization* - Die grafische Darstellung von Daten und Analyseergebnissen. Macht komplexe Zusammenhänge auf einen Blick verständlich.

*Pandas* - Die Standard-Bibliothek in Python für Datenmanipulation und -analyse. Bietet DataFrames als zentrale Datenstruktur.

*NumPy* - Grundlegende Bibliothek für numerische Berechnungen in Python. Basis für viele andere Data Science Tools.

*Matplotlib & Seaborn* - Die wichtigsten Visualisierungsbibliotheken in Python. Matplotlib ist die Basis, Seaborn bietet schönere, statistisch orientierte Plots.

*Plotly* - Interaktive Visualisierungsbibliothek, die moderne, webbasierte Grafiken ermöglicht.

*Missing Data* - Fehlende Werte in Datensätzen. Müssen behandelt werden (löschen, imputieren, etc.), bevor man analysieren kann.

*Outliers* - Ausreißer in den Daten, die stark vom Rest abweichen. Können echte Anomalien oder Messfehler sein.

*Data Normalization* - Das Skalieren von Daten auf einen einheitlichen Bereich (z.B. 0-1), damit verschiedene Features vergleichbar werden.

*Correlation Analysis* - Untersuchung von Zusammenhängen zwischen Variablen. Zeigt, welche Features miteinander in Beziehung stehen.

*Distribution* - Die Verteilung von Werten in einem Dataset. Wichtig zu verstehen, ob Daten normal-, schief- oder anders verteilt sind.

```python
# Setup: Installation der benötigten Bibliotheken
!pip install -q pandas numpy matplotlib seaborn plotly scikit-learn

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
import warnings

# Konfiguration
warnings.filterwarnings('ignore')
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print("Alle Bibliotheken erfolgreich geladen!")
print(f"Pandas Version: {pd.__version__}")
print(f"NumPy Version: {np.__version__}")
```

```python
# Beispiel-Dataset erstellen (simulierte Daten)
# In der Praxis würde man echte Daten laden: pd.read_csv('data.csv')

np.random.seed(42)

# Simulierte Verkaufsdaten eines Online-Shops
n_samples = 1000

data = {
    'TransactionID': range(1, n_samples + 1),
    'Date': pd.date_range('2024-01-01', periods=n_samples, freq='H'),
    'Product_Category': np.random.choice(['Electronics', 'Clothing', 'Food', 'Books', 'Sports'], n_samples),
    'Price': np.random.uniform(5, 500, n_samples).round(2),
    'Quantity': np.random.randint(1, 10, n_samples),
    'Customer_Age': np.random.normal(35, 15, n_samples).round(0),
    'Rating': np.random.choice([1, 2, 3, 4, 5, np.nan], n_samples, p=[0.05, 0.1, 0.15, 0.35, 0.3, 0.05]),
    'Discount': np.random.choice([0, 5, 10, 15, 20], n_samples, p=[0.4, 0.2, 0.2, 0.15, 0.05])
}

df = pd.DataFrame(data)
df['Revenue'] = df['Price'] * df['Quantity'] * (1 - df['Discount']/100)

print("Dataset erfolgreich erstellt!")
print(f"\nDataset Info:")
print(f"   - Anzahl Zeilen: {len(df)}")
print(f"   - Anzahl Spalten: {len(df.columns)}")
print(f"\nErste 5 Zeilen:")
df.head()
```

```python
# Exploratory Data Analysis (EDA): Erste Inspektion

print("=" * 60)
print("DATASET ÜBERSICHT")
print("=" * 60)

# Grundlegende Informationen
print("\n1. Datentypen und fehlende Werte:")
print(df.info())

print("\n2. Statistische Zusammenfassung:")
print(df.describe())

print("\n3. Fehlende Werte pro Spalte:")
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({
    'Fehlende Werte': missing,
    'Prozent': missing_pct
})
print(missing_df[missing_df['Fehlende Werte'] > 0])

print("\n4. Kategoriale Variablen - Verteilung:")
print(df['Product_Category'].value_counts())

print("\n5. Numerische Statistiken:")
print(f"   - Durchschnittlicher Preis: €{df['Price'].mean():.2f}")
print(f"   - Durchschnittlicher Umsatz: €{df['Revenue'].mean():.2f}")
print(f"   - Gesamtumsatz: €{df['Revenue'].sum():.2f}")
```

```python
# Data Preprocessing: Datenbereinigung und -transformation

print("DATA PREPROCESSING")
print("=" * 60)

# 1. Fehlende Werte behandeln
print("\n1. Behandlung fehlender Werte...")
print(f"   Fehlende Ratings vor Imputation: {df['Rating'].isnull().sum()}")

# Rating mit Median imputieren
imputer = SimpleImputer(strategy='median')
df['Rating'] = imputer.fit_transform(df[['Rating']])

print(f"   Fehlende Ratings nach Imputation: {df['Rating'].isnull().sum()}")

# 2. Outlier Detection
print("\n2. Outlier-Detektion für Customer_Age...")
Q1 = df['Customer_Age'].quantile(0.25)
Q3 = df['Customer_Age'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Customer_Age'] < lower_bound) | (df['Customer_Age'] > upper_bound)]
print(f"   Gefundene Outliers: {len(outliers)}")
print(f"   Bereich: {lower_bound:.0f} - {upper_bound:.0f} Jahre")

# Outliers bereinigen (nur realistische Werte behalten)
df_clean = df[(df['Customer_Age'] >= 18) & (df['Customer_Age'] <= 80)].copy()
print(f"   Zeilen nach Bereinigung: {len(df_clean)}")

# 3. Feature Engineering
print("\n3. Feature Engineering...")
df_clean['Month'] = df_clean['Date'].dt.month
df_clean['Day_of_Week'] = df_clean['Date'].dt.dayofweek
df_clean['Hour'] = df_clean['Date'].dt.hour
df_clean['Is_Weekend'] = df_clean['Day_of_Week'].isin([5, 6]).astype(int)
df_clean['Price_Category'] = pd.cut(df_clean['Price'], bins=[0, 50, 150, 500], labels=['Low', 'Medium', 'High'])

print("   Neue Features erstellt:")
print("   - Month, Day_of_Week, Hour")
print("   - Is_Weekend")
print("   - Price_Category")

# 4. Normalisierung (für ML vorbereiten)
print("\n4. Daten-Normalisierung...")
scaler = StandardScaler()
numerical_cols = ['Price', 'Quantity', 'Customer_Age', 'Revenue']
df_clean[['Price_scaled', 'Quantity_scaled', 'Age_scaled', 'Revenue_scaled']] = scaler.fit_transform(df_clean[numerical_cols])

print("   Numerische Features normalisiert (StandardScaler)")
print("\nPreprocessing abgeschlossen!")
print(f"Finale Dataset-Größe: {df_clean.shape}")
```

```python
# Visualisierung Teil 1: Verteilungen und Trends (Matplotlib & Seaborn)

print("DATENVISUALISIERUNG MIT MATPLOTLIB & SEABORN")
print("=" * 60)

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Verteilung des Umsatzes
axes[0, 0].hist(df_clean['Revenue'], bins=50, color='skyblue', edgecolor='black', alpha=0.7)
axes[0, 0].set_title('Verteilung des Umsatzes', fontsize=14, fontweight='bold')
axes[0, 0].set_xlabel('Umsatz (€)')
axes[0, 0].set_ylabel('Häufigkeit')
axes[0, 0].axvline(df_clean['Revenue'].mean(), color='red', linestyle='--', linewidth=2, label=f'Mittelwert: €{df_clean["Revenue"].mean():.2f}')
axes[0, 0].legend()

# 2. Umsatz nach Produktkategorie (Boxplot)
df_clean.boxplot(column='Revenue', by='Product_Category', ax=axes[0, 1])
axes[0, 1].set_title('Umsatz nach Produktkategorie', fontsize=14, fontweight='bold')
axes[0, 1].set_xlabel('Produktkategorie')
axes[0, 1].set_ylabel('Umsatz (€)')
plt.sca(axes[0, 1])
plt.xticks(rotation=45)

# 3. Korrelation zwischen numerischen Variablen (Heatmap)
correlation = df_clean[['Price', 'Quantity', 'Customer_Age', 'Rating', 'Revenue']].corr()
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8}, ax=axes[1, 0])
axes[1, 0].set_title('Korrelationsmatrix', fontsize=14, fontweight='bold')

# 4. Umsatz über Zeit (Trend)
daily_revenue = df_clean.groupby(df_clean['Date'].dt.date)['Revenue'].sum()
axes[1, 1].plot(daily_revenue.index, daily_revenue.values, linewidth=2, color='green')
axes[1, 1].set_title('Täglicher Umsatz-Trend', fontsize=14, fontweight='bold')
axes[1, 1].set_xlabel('Datum')
axes[1, 1].set_ylabel('Umsatz (€)')
axes[1, 1].tick_params(axis='x', rotation=45)
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\nStatische Visualisierungen erstellt!")
```

```python
# Visualisierung Teil 2: Interaktive Plots mit Plotly

print("INTERAKTIVE VISUALISIERUNG MIT PLOTLY")
print("=" * 60)

# 1. Interaktiver Scatter Plot: Preis vs Umsatz
fig1 = px.scatter(df_clean, 
                  x='Price', 
                  y='Revenue', 
                  color='Product_Category',
                  size='Quantity',
                  hover_data=['Customer_Age', 'Rating'],
                  title='Preis vs. Umsatz nach Produktkategorie',
                  labels={'Price': 'Preis (€)', 'Revenue': 'Umsatz (€)'},
                  template='plotly_white')
fig1.update_layout(height=500)
fig1.show()

# 2. Interaktives Balkendiagramm: Durchschnittlicher Umsatz nach Kategorie
avg_revenue = df_clean.groupby('Product_Category')['Revenue'].mean().sort_values(ascending=False)
fig2 = px.bar(x=avg_revenue.index, 
              y=avg_revenue.values,
              title='Durchschnittlicher Umsatz nach Produktkategorie',
              labels={'x': 'Produktkategorie', 'y': 'Durchschn. Umsatz (€)'},
              color=avg_revenue.values,
              color_continuous_scale='Viridis',
              template='plotly_white')
fig2.update_layout(showlegend=False, height=500)
fig2.show()

# 3. Zeitreihen-Plot: Umsatz über Stunden
hourly_revenue = df_clean.groupby('Hour')['Revenue'].agg(['sum', 'mean', 'count'])
fig3 = go.Figure()
fig3.add_trace(go.Scatter(x=hourly_revenue.index, y=hourly_revenue['sum'], 
                          mode='lines+markers', name='Gesamtumsatz',
                          line=dict(color='blue', width=3)))
fig3.add_trace(go.Scatter(x=hourly_revenue.index, y=hourly_revenue['mean'], 
                          mode='lines+markers', name='Durchschn. Umsatz',
                          line=dict(color='orange', width=2, dash='dash')))
fig3.update_layout(title='Umsatz nach Tageszeit',
                   xaxis_title='Stunde des Tages',
                   yaxis_title='Umsatz (€)',
                   template='plotly_white',
                   hovermode='x unified',
                   height=500)
fig3.show()

# 4. Sunburst Chart: Hierarchische Verteilung
category_counts = df_clean.groupby(['Product_Category', 'Price_Category']).size().reset_index(name='Count')
fig4 = px.sunburst(category_counts,
                   path=['Product_Category', 'Price_Category'],
                   values='Count',
                   title='Hierarchische Verteilung: Produkte nach Preis',
                   color='Count',
                   color_continuous_scale='RdYlGn',
                   template='plotly_white')
fig4.update_layout(height=600)
fig4.show()

print("\nInteraktive Visualisierungen erstellt!")
print("Tipp: Bewege die Maus über die Plots für Details, zoome und rotiere!")
```

```python
# KI-gestützte Insights: Automatische Analyse mit Gemini

print("KI-GESTÜTZTE DATENANALYSE MIT GEMINI")
print("=" * 60)

# Gemini API Setup
# 
# SO RICHTEST DU DEN API-KEY EIN:
# 1. Gehe zu https://aistudio.google.com/app/apikey und erstelle einen kostenlosen API-Key
# 2. In Google Colab: Klicke auf das Schlüssel-Symbol in der linken Sidebar
# 3. Klicke auf "+ Add new secret"
# 4. Name: GOOGLE_API_KEY
# 5. Value: Füge deinen API-Key ein
# 6. Aktiviere "Notebook access"
# 
# Der Code lädt dann automatisch den Key aus den Colab Secrets

try:
    import google.generativeai as genai
    from google.colab import userdata
    
    # API-Key aus Colab Secrets laden
    GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
    genai.configure(api_key=GOOGLE_API_KEY)
    
    # Modell initialisieren: gemini-1.5-flash
    model = genai.GenerativeModel('gemini-1.5-flash')
    print("Gemini API konfiguriert (gemini-1.5-flash)")
    
except Exception as e:
    print(f"Warnung: Gemini API nicht verfügbar - {e}")
    print("Fallback auf manuelle Analyse...\n")
    model = None

# Zusammenfassung der wichtigsten Statistiken für das LLM
summary_stats = f"""
Analysiere bitte die folgenden E-Commerce Daten und generiere actionable Insights:

DATASET ÜBERSICHT:
- Gesamtanzahl Transaktionen: {len(df_clean)}
- Zeitraum: {df_clean['Date'].min()} bis {df_clean['Date'].max()}

UMSATZ-STATISTIKEN:
- Gesamtumsatz: €{df_clean['Revenue'].sum():.2f}
- Durchschnittlicher Umsatz pro Transaktion: €{df_clean['Revenue'].mean():.2f}
- Median Umsatz: €{df_clean['Revenue'].median():.2f}

TOP PRODUKTKATEGORIEN (nach Umsatz):
{df_clean.groupby('Product_Category')['Revenue'].sum().sort_values(ascending=False).to_dict()}

KUNDENDEMOGRAPHIE:
- Durchschnittsalter: {df_clean['Customer_Age'].mean():.1f} Jahre
- Altersbereich: {df_clean['Customer_Age'].min():.0f} - {df_clean['Customer_Age'].max():.0f} Jahre

BEWERTUNGEN:
- Durchschnittliche Bewertung: {df_clean['Rating'].mean():.2f}/5
- Verteilung: {df_clean['Rating'].value_counts().to_dict()}

ZEITLICHE MUSTER:
- Umsatzstärkste Stunde: {df_clean.groupby('Hour')['Revenue'].sum().idxmax()}:00 Uhr
- Weekend vs Weekday Umsatz: Weekend: €{df_clean[df_clean['Is_Weekend']==1]['Revenue'].sum():.2f}, Weekday: €{df_clean[df_clean['Is_Weekend']==0]['Revenue'].sum():.2f}

KORRELATIONEN:
{df_clean[['Price', 'Quantity', 'Rating', 'Customer_Age', 'Revenue']].corr()['Revenue'].to_dict()}

AUFGABE:
Bitte analysiere diese Daten und liefere:
1. Die 3 wichtigsten Insights aus geschäftlicher Sicht
2. Konkrete Handlungsempfehlungen für Marketing und Vertrieb
3. Potenzielle Risiken oder Probleme, die adressiert werden sollten
4. Vorschläge für weitere Analysen oder A/B-Tests

Strukturiere deine Antwort klar und actionable.
"""

print("\n" + "="*60)
print("DATEN-SUMMARY FÜR KI-ANALYSE:")
print("="*60)
print(summary_stats)

# KI-Analyse durchführen
if model:
    print("\n" + "="*60)
    print("GEMINI KI-ANALYSE:")
    print("="*60)
    
    try:
        response = model.generate_content(summary_stats)
        ai_insights = response.text
        
        print("\n" + ai_insights)
        
        print("\n" + "="*60)
        print("KI-Analyse erfolgreich abgeschlossen!")
        print("="*60)
        
    except Exception as e:
        error_msg = str(e)
        print(f"\nFehler bei der KI-Analyse: {error_msg[:150]}...")
        print("Bitte API-Key und Modellverfügbarkeit prüfen.\n")
```
