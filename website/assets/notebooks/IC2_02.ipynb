**Begriffe, die immer wieder auftauchen**:

*Data Preprocessing* - Die Aufbereitung von Rohdaten für die Analyse. Umfasst Bereinigung, Transformation, Normalisierung und das Handling von fehlenden Werten.

*Exploratory Data Analysis (EDA)* - Der erste Schritt jeder Datenanalyse. Dabei werden Muster, Anomalien und Zusammenhänge in den Daten entdeckt, meist visuell.

*Feature Engineering* - Das Erstellen neuer, aussagekräftiger Features aus vorhandenen Daten, um die Modellperformance zu verbessern.

*Data Visualization* - Die grafische Darstellung von Daten und Analyseergebnissen. Macht komplexe Zusammenhänge auf einen Blick verständlich.

*Pandas* - Die Standard-Bibliothek in Python für Datenmanipulation und -analyse. Bietet DataFrames als zentrale Datenstruktur.

*NumPy* - Grundlegende Bibliothek für numerische Berechnungen in Python. Basis für viele andere Data Science Tools.

*Matplotlib & Seaborn* - Die wichtigsten Visualisierungsbibliotheken in Python. Matplotlib ist die Basis, Seaborn bietet schönere, statistisch orientierte Plots.

*Plotly* - Interaktive Visualisierungsbibliothek, die moderne, webbasierte Grafiken ermöglicht.

*Missing Data* - Fehlende Werte in Datensätzen. Müssen behandelt werden (löschen, imputieren, etc.), bevor man analysieren kann.

*Outliers* - Ausreißer in den Daten, die stark vom Rest abweichen. Können echte Anomalien oder Messfehler sein.

*Data Normalization* - Das Skalieren von Daten auf einen einheitlichen Bereich (z.B. 0-1), damit verschiedene Features vergleichbar werden.

*Correlation Analysis* - Untersuchung von Zusammenhängen zwischen Variablen. Zeigt, welche Features miteinander in Beziehung stehen.

*Distribution* - Die Verteilung von Werten in einem Dataset. Wichtig zu verstehen, ob Daten normal-, schief- oder anders verteilt sind.

```python
# Setup: Installation der benötigten Bibliotheken
!pip install -q pandas numpy matplotlib seaborn plotly scikit-learn

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
import warnings

# Konfiguration
warnings.filterwarnings('ignore')
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print("Alle Bibliotheken erfolgreich geladen!")
print(f"Pandas Version: {pd.__version__}")
print(f"NumPy Version: {np.__version__}")
```

```python
# Beispiel-Dataset erstellen (simulierte Daten)
# In der Praxis würde man echte Daten laden: pd.read_csv('data.csv')

np.random.seed(42)

# Simulierte Verkaufsdaten eines Online-Shops
n_samples = 1000

data = {
    'TransactionID': range(1, n_samples + 1),
    'Date': pd.date_range('2024-01-01', periods=n_samples, freq='H'),
    'Product_Category': np.random.choice(['Electronics', 'Clothing', 'Food', 'Books', 'Sports'], n_samples),
    'Price': np.random.uniform(5, 500, n_samples).round(2),
    'Quantity': np.random.randint(1, 10, n_samples),
    'Customer_Age': np.random.normal(35, 15, n_samples).round(0),
    'Rating': np.random.choice([1, 2, 3, 4, 5, np.nan], n_samples, p=[0.05, 0.1, 0.15, 0.35, 0.3, 0.05]),
    'Discount': np.random.choice([0, 5, 10, 15, 20], n_samples, p=[0.4, 0.2, 0.2, 0.15, 0.05])
}

df = pd.DataFrame(data)
df['Revenue'] = df['Price'] * df['Quantity'] * (1 - df['Discount']/100)

print("Dataset erfolgreich erstellt!")
print(f"\nDataset Info:")
print(f"   - Anzahl Zeilen: {len(df)}")
print(f"   - Anzahl Spalten: {len(df.columns)}")
print(f"\nErste 5 Zeilen:")
df.head()
```

```python
# Schnelle Datenübersicht

print("DATASET ÜBERSICHT")
print("=" * 60)

print("\n1. Grundlegende Informationen:")
print(f"   - Anzahl Zeilen: {len(df)}")
print(f"   - Anzahl Spalten: {len(df.columns)}")
print(f"   - Fehlende Werte: {df.isnull().sum().sum()}")

print("\n2. Key Statistiken:")
print(f"   - Gesamtumsatz: €{df['Revenue'].sum():.2f}")
print(f"   - Durchschnittlicher Umsatz: €{df['Revenue'].mean():.2f}")
print(f"   - Top Kategorie: {df.groupby('Product_Category')['Revenue'].sum().idxmax()}")

print("\nErste Zeilen:")
df.head()
```

```python
# Minimales Preprocessing - Fokus auf KI-Analyse vorbereiten

print("DATA PREPROCESSING (vereinfacht)")
print("=" * 60)

# 1. Fehlende Werte behandeln
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='median')
df['Rating'] = imputer.fit_transform(df[['Rating']])
print("Fehlende Ratings imputiert")

# 2. Bereinigung: Nur realistische Werte
df_clean = df[(df['Customer_Age'] >= 18) & (df['Customer_Age'] <= 80)].copy()
print(f"Zeilen nach Bereinigung: {len(df_clean)}")

# 3. Zeitliche Features für Analyse
df_clean['Month'] = df_clean['Date'].dt.month
df_clean['Hour'] = df_clean['Date'].dt.hour
df_clean['Is_Weekend'] = df_clean['Date'].dt.dayofweek.isin([5, 6]).astype(int)

print("Features erstellt: Month, Hour, Is_Weekend")
print(f"\nDaten bereit für KI-Analyse: {df_clean.shape}")
```

```python
# Schnelle Visualisierung - Überblick für KI-Kontext

import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Umsatz nach Kategorie
category_revenue = df_clean.groupby('Product_Category')['Revenue'].sum().sort_values()
axes[0].barh(category_revenue.index, category_revenue.values, color='skyblue')
axes[0].set_title('Umsatz nach Kategorie', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Umsatz (€)')

# Zeitlicher Trend
daily_revenue = df_clean.groupby(df_clean['Date'].dt.date)['Revenue'].sum()
axes[1].plot(daily_revenue.index, daily_revenue.values, linewidth=2, color='green')
axes[1].set_title('Täglicher Umsatz', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Datum')
axes[1].set_ylabel('Umsatz (€)')
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Visualisierungen erstellt - Kontext für KI-Analyse")
```

```python

# KI-gestützte Insights: Automatische Analyse mit Gemini

print("KI-GESTÜTZTE DATENANALYSE MIT GEMINI")
print("=" * 60)

# Gemini API Setup
# 
# SO RICHTEST DU DEN API-KEY EIN:
# 1. Gehe zu https://aistudio.google.com/app/apikey und erstelle einen kostenlosen API-Key
# 2. In Google Colab: Klicke auf das Schlüssel-Symbol in der linken Sidebar
# 3. Klicke auf "+ Add new secret"
# 4. Name: GOOGLE_API_KEY
# 5. Value: Füge deinen API-Key ein
# 6. Aktiviere "Notebook access"
# 
# Der Code lädt dann automatisch den Key aus den Colab Secrets

try:
    from google import genai
    from google.colab import userdata
    
    # API-Key aus Colab Secrets laden
    GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
    client = genai.Client(api_key=GOOGLE_API_KEY)
    print("Gemini API konfiguriert (gemini-2.0-flash)")
    
except Exception as e:
    print(f"Warnung: Gemini API nicht verfügbar - {e}")
    print("Fallback auf manuelle Analyse...\n")
    client = None

# Zusammenfassung der wichtigsten Statistiken für das LLM
summary_stats = f"""
Analysiere bitte die folgenden E-Commerce Daten und generiere actionable Insights:

DATASET ÜBERSICHT:
- Gesamtanzahl Transaktionen: {len(df_clean)}
- Zeitraum: {df_clean['Date'].min()} bis {df_clean['Date'].max()}

UMSATZ-STATISTIKEN:
- Gesamtumsatz: €{df_clean['Revenue'].sum():.2f}
- Durchschnittlicher Umsatz pro Transaktion: €{df_clean['Revenue'].mean():.2f}
- Median Umsatz: €{df_clean['Revenue'].median():.2f}

TOP PRODUKTKATEGORIEN (nach Umsatz):
{df_clean.groupby('Product_Category')['Revenue'].sum().sort_values(ascending=False).to_dict()}

KUNDENDEMOGRAPHIE:
- Durchschnittsalter: {df_clean['Customer_Age'].mean():.1f} Jahre
- Altersbereich: {df_clean['Customer_Age'].min():.0f} - {df_clean['Customer_Age'].max():.0f} Jahre

BEWERTUNGEN:
- Durchschnittliche Bewertung: {df_clean['Rating'].mean():.2f}/5
- Verteilung: {df_clean['Rating'].value_counts().to_dict()}

ZEITLICHE MUSTER:
- Umsatzstärkste Stunde: {df_clean.groupby('Hour')['Revenue'].sum().idxmax()}:00 Uhr
- Weekend vs Weekday Umsatz: Weekend: €{df_clean[df_clean['Is_Weekend']==1]['Revenue'].sum():.2f}, Weekday: €{df_clean[df_clean['Is_Weekend']==0]['Revenue'].sum():.2f}

KORRELATIONEN:
{df_clean[['Price', 'Quantity', 'Rating', 'Customer_Age', 'Revenue']].corr()['Revenue'].to_dict()}

AUFGABE:
Bitte analysiere diese Daten und liefere:
1. Die 3 wichtigsten Insights aus geschäftlicher Sicht
2. Konkrete Handlungsempfehlungen für Marketing und Vertrieb
3. Potenzielle Risiken oder Probleme, die adressiert werden sollten
4. Vorschläge für weitere Analysen oder A/B-Tests

Strukturiere deine Antwort klar und actionable.
"""

print("\n" + "="*60)
print("DATEN-SUMMARY FÜR KI-ANALYSE:")
print("="*60)
print(summary_stats)

# KI-Analyse durchführen
if client:
    print("\n" + "="*60)
    print("GEMINI KI-ANALYSE:")
    print("="*60)
    
    try:
        response = client.models.generate_content(
            model='gemini-2.0-flash',
            contents=summary_stats
        )
        ai_insights = response.text
        
        print("\n" + ai_insights)
        
        print("\n" + "="*60)
        print("KI-Analyse erfolgreich abgeschlossen!")
        print("="*60)
        
    except Exception as e:
        error_msg = str(e)
        print(f"\nFehler bei der KI-Analyse: {error_msg[:150]}...")
        print("Bitte API-Key und Modellverfügbarkeit prüfen.\n")
```
