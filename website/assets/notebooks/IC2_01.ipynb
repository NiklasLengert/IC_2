**Begriffe, die immer wieder auftauchen**:

*Tokenisierung* - Texte werden in kleinere Einheiten (Tokens) zerlegt – also Wörter, Teilwörter oder einzelne Zeichen. Nur so kann der Computer damit arbeiten.

*NLP* - Natural Language Processing. Ein Feld in der KI, das sich damit beschäftigt, wie Computer menschliche Sprache verstehen und verarbeiten können.

*NER* - Named Entity Recognition. Dabei werden bestimmte Entitäten im Text erkannt – z.B. Personennamen, Orte, Organisationen oder Zeitangaben.

*KI* - Künstliche Intelligenz. Programme, die versuchen, menschliche Denkprozesse nachzuahmen.

*spaCy* - Industriestärke NLP-Bibliothek in Python. Schnell, effizient und mit vortrainierten Modellen für viele Sprachen.

*LLM* - Large Language Model. Große Sprachmodelle wie GPT, Claude oder Gemini, die auf riesigen Textmengen trainiert wurden.

*Transformer* - Eine Netzwerk-Architektur von Google aus 2017. Nutzt Attention-Mechanismen, um den Kontext aller Wörter gleichzeitig zu erfassen – daher so effektiv.

*Schlüsselwort- und Phrasenextraktion* - Die wichtigsten Begriffe aus einem Text automatisch rausfiltern, um schnell zu erfassen, worum es geht.

*Part of Speech (POS) Tagging* - Jedem Wort wird seine grammatikalische Rolle zugewiesen (Nomen, Verb, Adjektiv usw.), je nach Kontext im Satz.

*Gradio* - Python-Framework zum schnellen Erstellen von interaktiven UIs für ML-Modelle und Prototypen.

*API* - Application Programming Interface. Schnittstelle, über die man externe Dienste (wie LLMs) in eigene Programme einbinden kann.

```python
# Setup: Installation der benötigten Pakete
# Gradio für die UI, spaCy für NLP, das deutsche Sprachmodell und Gemini für LLM-Integration
!pip install -q gradio
!pip install -q spacy
!pip install -q google-genai
!python -m spacy download de_core_news_lg

import gradio as gr
import spacy
from google import genai

# spaCy Modell laden
nlp = spacy.load("de_core_news_lg")

# Gemini Client initialisieren (falls in Colab Secrets verfügbar)
client = None
try:
    from google.colab import userdata
    GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
    client = genai.Client(api_key=GOOGLE_API_KEY)
    print("Gemini API erfolgreich konfiguriert")
except Exception as e:
    print(f"API-Schlüssel nicht gefunden: {e}")
    print("Tipp: API-Schlüssel im Colab Secrets Manager als 'GOOGLE_API_KEY' hinterlegen.")

print("Setup abgeschlossen!")
```

```python

# NLP-Pipeline: Erweiterte Analyse mit Keyphrases und Named Entities
import time
import re

def analyze_text_with_nlp(text):
    """
    Analysiert den Text mit spaCy und extrahiert:
    - Schlüsselphrasen (Keyphrases)
    - Named Entities (Personen, Orte, Organisationen, etc.)
    - Part-of-Speech Tags für wichtige Begriffe
    """
    doc = nlp(text)
    
    # 1. Schlüsselphrasen extrahieren
    key_phrases = set()
    spacy_stopwords = nlp.Defaults.stop_words
    
    # Einzelne relevante Nomen und Eigennamen sammeln
    for token in doc:
        token_text_lower = token.text.lower()
        if token_text_lower in spacy_stopwords:
            continue
        if token.pos_ in ['NOUN', 'PROPN'] and len(token.text) > 3:
            key_phrases.add(token.text)
    
    # Längere Phrasen (noun chunks) hinzufügen
    for chunk in doc.noun_chunks:
        phrase = chunk.text
        # Artikel am Anfang entfernen
        if phrase.lower().split()[0] in ['der', 'die', 'das', 'ein', 'eine']:
            phrase = " ".join(phrase.split()[1:])
        if len(phrase.split()) > 1:
            key_phrases.add(phrase)
    
    keyphrases_list = sorted(list(key_phrases), key=len, reverse=True)
    
    # 2. Named Entities extrahieren
    entities = {}
    for ent in doc.ents:
        ent_type = ent.label_
        if ent_type not in entities:
            entities[ent_type] = []
        entities[ent_type].append(ent.text)
    
    # Ergebnisse formatieren
    output = "## NLP-Analyse Ergebnisse\n\n"
    
    output += f"### Schlüsselphrasen ({len(keyphrases_list)} gefunden):\n"
    output += ", ".join(keyphrases_list[:20]) + "\n\n"  # Top 20 anzeigen
    
    if entities:
        output += "### Named Entities:\n"
        for ent_type, ent_list in entities.items():
            output += f"- **{ent_type}**: {', '.join(set(ent_list))}\n"
    else:
        output += "### Named Entities:\n"
        output += "Keine spezifischen Entitäten erkannt.\n"
    
    return output, keyphrases_list, entities


import time
import re

def process_with_gemini(keyphrases, entities, original_text):
    """
    Sendet die extrahierten Informationen an Gemini für semantische Analyse.
    Behandelt Rate-Limits proaktiv und nutzt saubere Retries.
    """
    prompt = f"""Analysiere folgende Forschungsanfrage und hilf bei der Literatursuche:

**Original-Text:**
{original_text}

**Extrahierte Schlüsselbegriffe:**
{', '.join(keyphrases[:15])}

**Erkannte Entitäten:**
{entities}

Bitte:
1. Fasse das Hauptthema in 1-2 Sätzen zusammen
2. Identifiziere die 3-5 wichtigsten Suchbegriffe für eine akademische Literatursuche
3. Schlage verwandte Forschungsgebiete oder Konzepte vor, die relevant sein könnten
4. Empfehle spezifische Datenbanken (z.B. arXiv, PubMed, IEEE Xplore), die für diese Anfrage geeignet sind

Antworte strukturiert und präzise."""

    models_to_try = ['gemini-2.5-flash', 'gemini-2.5-flash-lite']

    for model in models_to_try:
        max_retries = 3
        
        for attempt in range(max_retries):
            try:
                # API-Aufruf (Syntax für das google-genai SDK)
                response = client.models.generate_content(model=model, contents=prompt)
                
                # PROAKTIVES WARTEN: Verhindert, dass das Free-Tier Limit (15 RPM) überhaupt erreicht wird
                time.sleep(4) 
                
                model_note = f" *(Modell: {model})*" if model != 'gemini-2.5-flash' else ""
                return f"\n## KI-Analyse (Gemini){model_note}\n\n{response.text}"
                
            except Exception as e:
                error_str = str(e)
                
                # Wenn das Rate Limit doch zuschlägt
                if '429' in error_str or 'RESOURCE_EXHAUSTED' in error_str:
                    print(f"Rate Limit erreicht bei {model} (Versuch {attempt + 1}/{max_retries}). Warte...")
                    match = re.search(r'retryDelay.*?(\d+)s', error_str)
                    wait_time = int(match.group(1)) + 2 if match else 35
                    time.sleep(wait_time)
                    continue # Versucht denselben API-Aufruf in der Schleife noch einmal
                    
                # Wenn das Modell nicht existiert (404) oder die Region es nicht unterstützt
                elif '404' in error_str:
                    print(f"Modell {model} nicht gefunden (404). Wechsle zum Fallback-Modell...")
                    break # Bricht die Retry-Schleife ab und geht zum nächsten Modell in models_to_try
                    
                # Bei anderen unvorhergesehenen Fehlern
                else:
                    return f"\n## KI-Analyse nicht verfügbar\n\nFehler: {e}\n\nTipp: API-Schlüssel oder Internetverbindung prüfen."

    return (
        "\n## KI-Analyse nicht verfügbar\n\n"
        "Das Free-Tier-Kontingent ist erschöpft oder die Modelle sind nicht erreichbar. "
        "Bitte warte einige Minuten und versuche es erneut."
    )


def full_pipeline(text):
    """
    Vollständiger Workflow: NLP-Analyse → LLM-Integration → Strukturierte Ausgabe
    """
    if not text or len(text.strip()) < 10:
        return "Bitte gib einen längeren Text ein (mindestens 10 Zeichen)."
    
    # Schritt 1: NLP-Analyse
    nlp_output, keyphrases, entities = analyze_text_with_nlp(text)
    
    # Schritt 2: Gemini-Analyse
    gemini_output = process_with_gemini(keyphrases, entities, text)
    
    # Schritt 3: Kombinierte, strukturierte Ausgabe
    final_output = nlp_output + gemini_output
    
    # Schritt 4: Placeholder für zukünftige Literatursuche
    final_output += "\n\n## Literatursuche (In Entwicklung)\n\n"
    final_output += "*Hier werden zukünftig automatisch Paper-Vorschläge von arXiv, PubMed und Google Scholar erscheinen.*\n"
    final_output += f"*Suchbegriffe für manuelle Suche:* {', '.join(keyphrases[:5])}"
    
    return final_output


# Gradio Interface erstellen
with gr.Blocks(title="Intelligente Literatursuche", theme=gr.themes.Soft()) as demo:
    gr.Markdown("# Intelligente Literatursuche mit NLP & LLM")
    gr.Markdown("Gib eine Forschungsfrage oder Themenbeschreibung ein, und das System extrahiert automatisch relevante Begriffe und gibt Empfehlungen für die Literatursuche.")
    
    with gr.Row():
        with gr.Column():
            input_text = gr.Textbox(
                label="Deine Forschungsfrage / Thema",
                placeholder="Beispiel: Wie wirkt sich maschinelles Lernen auf die medizinische Diagnostik aus? Welche neuronalen Netze werden für Bilderkennung von Krankheiten verwendet?",
                lines=8
            )
            analyze_btn = gr.Button("Analysieren", variant="primary")
        
        with gr.Column():
            output_text = gr.Textbox(
                label="Analyse-Ergebnisse",
                lines=20
            )
    
    analyze_btn.click(fn=full_pipeline, inputs=input_text, outputs=output_text)
    
    gr.Markdown("---")
    gr.Markdown("**Tipp:** Je detaillierter deine Anfrage, desto bessere Ergebnisse erhältst du!")

demo.launch(share=True, debug=True)
```
